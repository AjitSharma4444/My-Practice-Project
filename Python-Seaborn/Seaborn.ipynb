# 🎨 Why the Heatmaps Look Different
Both heatmaps show the same underlying data (the correlation coefficients of the mtcars variables), but they use different visual scales:
1. Color Maps (CMAP)
* Previous Heatmap: Used a divergent color map (likely 'RdBu' or similar) where red/orange signified positive correlation, and dark purple/black signified negative correlation.
* Current Heatmap: Uses a different color map (likely 'magma', 'plasma', or a custom scale). Here:
   * Bright Red/Orange represents Positive correlation.
   * Dark Purple/Black represents Negative correlation.
   * The transition colors (pinks, lighter purples) are near zero.
2. Scale Range
The color bar in the new image ranges from -0.8 to +0.8, with ticks at -0.4, 0.0, and 0.4. The previous heatmap had a range that extended to +1.0 and -1.0.
Despite the visual differences, if you look at the pattern of dark (negative correlation) and light/red (positive correlation) squares, they match the expected relationships of the mtcars variables:
* mpg is dark (negatively correlated) with cyl, disp, and wt.
* hp is dark (negatively correlated) with qsec.
* gear and am (Automatic/Manual transmission) show strong positive correlation with each other (light squares in the lower right).
* In summary, the correlation calculation itself is consistent; only the visualization style has changed, which is controlled by the plotting function's arguments (e.g., the cmap parameter in seaborn.heatmap).
