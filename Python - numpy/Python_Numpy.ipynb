# Numpy space occupancy comparision with python_array

```python
import sys
```

```python
py_ar = [0, 4, 55, 2]
```

```python
np_ar = np.array(py_ar)
```

```python
sys.getsizeof(1) * len(py_ar) # python array size occupancy
```

```python
np_ar.itemsize * np_ar.size # numpy array size occupancy
```

```python
(32/112)*100 # less than 1/3 is occupied by numpy
```
